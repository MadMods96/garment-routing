# Results

This notebook displays `reports/results.json`. It does not recompute scores and it does not type them in.
Temperature scaling and the abstention threshold were fit on the validation split inside `garmentroute.evaluate`.


```python
import json
from pathlib import Path
from IPython.display import Image, display, Markdown
from garmentroute.reporting import render_results

results = json.loads(Path("reports/results.json").read_text(encoding="utf-8"))
eda_path = Path("reports/eda_summary.json")
eda = json.loads(eda_path.read_text(encoding="utf-8")) if eda_path.exists() else None
display(Markdown(render_results(results, eda)))
for path in sorted(Path("reports/figures").glob("coverage_*.png")):
    print(path)
    display(Image(filename=str(path)))
for path in sorted(Path("reports/figures").glob("reliability_*.png")):
    print(path)
    display(Image(filename=str(path)))
```

```python
mistakes = (((results.get("fusion") or {}).get("splits") or {}).get("test") or {}).get("after_temperature", {}).get("worst_mistakes")
print(json.dumps(mistakes, indent=2) if mistakes else "No scored mistakes yet.")
```
