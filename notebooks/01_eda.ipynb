# Exploratory look at the second-hand clothing index

Counts and figures come from `data/index.parquet`. Nothing in this notebook types a class count by hand.
Sample photographs appear only after the front images have been downloaded.


```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display
from garmentroute.eda import write_eda

summary = write_eda(Path("data/index.parquet"), Path("reports/figures"), Path("reports/eda_summary.json"), Path("data/raw"))
print("rows", summary["n_rows"], "gold", summary["n_gold"])
print("raw usage", json.dumps(summary["usage_raw_counts"], indent=2))
print("condition/usage correlation ratio", summary["condition_usage_correlation_ratio"])
frame = pd.read_parquet("data/index.parquet")
frame.groupby("station")["usage"].value_counts(dropna=False).head(30)
```

```python
for name in ("usage_counts.png", "usage_by_station.png", "missingness.png", "condition_by_usage.png"):
    path = Path("reports/figures") / name
    print(path)
    display(Image(filename=str(path)))
```
